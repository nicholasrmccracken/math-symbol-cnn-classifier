```python
import os
import torch
from torchvision import transforms
from PIL import Image
import numpy as np

parent = "Data\\train"

convert_tensor = transforms.Compose([
    transforms.ToTensor(),
    transforms.Resize((150, 135))
])

images = []
labels = []

label_map = {}
current_label = 0

for symbol in os.listdir(parent):
    full_path = os.path.join(parent, symbol)

    if symbol not in label_map:
        label_map[symbol] = current_label
        current_label += 1

    print(f"converting {symbol}s")

    for image in os.listdir(full_path):
        img_path = os.path.join(full_path, image)
        pil_image = Image.open(img_path).convert('L')
        image_tensor = convert_tensor(pil_image)
        images.append(image_tensor)
        labels.append(label_map[symbol])

X = torch.stack(images)
Y = torch.tensor(labels)

np.save("Datasets/X_train.npy", X.numpy())
np.save("Datasets/Y_train.npy", Y.numpy())
```
